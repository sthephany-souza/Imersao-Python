# Aula 1 - Análise de dados com pandas

```python
# importar biblioteca pandas
import pandas as pd
```

```python
# importar a base de dados
df = pd.read_csv("https://raw.githubusercontent.com/guilhermeonrails/data-jobs/refs/heads/main/salaries.csv")
```

```python
# mostrar a base de dados
df.head()
```

```python
# mostrar infos gerais da base de dados
df.info()
```

```python
# descrever, de maneira etalhada, a base
df.describe() # mostra dados numéricos
```

```python
# mostrar a quantidade de linhas e colunas da base e dados
df.shape

#(133349, 11) - (linhas, colunas)
```

```python
# mostrar a quantidade de linhas e colunas de outra maneira
linhas, colunas = df.shape[0], df.shape[1] # salva as infos nas variáveis

# mostra os valores das variáveis
print("linhas:", linhas)
print("colunas:", colunas)
```

```python
# listar as colunas
df.columns
```

```python
renomear_colunas = {
    'work_year': 'ano',
    'experience_level': 'senioridade',
    'employment_type': 'contrato',
    'job_title': 'cargo',
    'salary': 'salario',
    'salary_currency': 'moeda',
    'salary_in_usd': 'usd',
    'employee_residence': 'residencia',
    'remote_ratio': 'remoto',
    'company_location': 'empresa',
    'company_size': 'tamanho_empresa'
}

df.rename(columns=renomear_colunas, inplace=True)

df.columns
```

```python
# analisar colunas, parte por parte
df["senioridade"].value_counts()

# SE - Senior, MI - Pleno, EN - Junior, EX - Executivo
```

```python
df["contrato"].value_counts()

# FT - Full Time/Tempo Integral, CT - Contrato Temporário, PT - Part Time/Meio Períudo, FL - Freelancer
```

```python
df["remoto"].value_counts()

# 0 - Presenciais, 50 - Híbridas, 100 - Remotas
```

```python
df["tamanho_empresa"].value_counts()

# M - Média, L - Grande, S - Pequena
```

```python
# renomear atributos das colunas
renomear_senioridade = {
    'SE': 'Sênior',
    'MI': 'Pleno',
    'EN': 'Júnior',
    'EX': 'Executivo'
}

# adicionar mudança
df['senioridade'] = df['senioridade'].replace(renomear_senioridade)

# mostrar mudança
df['senioridade'].value_counts()
```

```python
renomear_contratos = {
    'FT': 'Tempo Integral',
    'CT': 'Contrato Temporário',
    'PT': 'Meio Período',
    'FL': 'Freelancer'
}

df['contrato'] = df['contrato'].replace(renomear_contratos)

df['contrato'].value_counts()
```

```python
renomear_tamanho_empresa = {
    'M': 'Média',
    'L': 'Grande',
    'S': 'Pequena'
}

df['tamanho_empresa'] = df['tamanho_empresa'].replace(renomear_tamanho_empresa)

df['tamanho_empresa'].value_counts()
```

```python
renomear_remoto = {
    0: 'Presencial',
    50: 'Híbrido',
    100: 'Remoto'
}

df['remoto'] = df['remoto'].replace(renomear_remoto)

df['remoto'].value_counts()
```

```python
# analise detalhada
df.describe(include=object) # com parâmetros mostra também string, não apenas number
```

```python
import numpy as np
```

```python
# criar um DataFrame de exemplo
df_salarios = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carlos", "Daniele", "Val"],
    "salario": [4000, np.nan, 5000, np.nan, 100000]
})

# calcular a média e substitui os valores nulos pela média
df_salarios["salario_media"] = df_salarios["salario"].fillna(df_salarios["salario"].mean().round(2)) # round(2) arredonda os valores

# calcular a mediana e substitui os valores nulos pela mediana

df_salarios["salario_mediana"] = df_salarios["salario"].fillna(df_salarios["salario"].median())

df_salarios

# ''' ''' - faz comentário de multiplas linhas
```

```python
df_temperaturas = pd.DataFrame({
    "dia": ["Segunda", "Terça", "Quarta", "Quinta", "Sexta"],
    "temperatura": [30, np.nan, np.nan, 28, 27]
})

df_temperaturas["preenchido_ffill"] = df_temperaturas["temperatura"].ffill() # forward fill -  preenche o valor nulo com o valor anterior

df_temperaturas
```

```python
df_temperaturas = pd.DataFrame({
    "dia": ["Segunda", "Terça", "Quarta", "Quinta", "Sexta"],
    "temperatura": [30, np.nan, np.nan, 28, 27]
})

df_temperaturas["preenchido_bfill"] = df_temperaturas["temperatura"].bfill() # back fill -  preenche o valor nulo com o valor seguinte

df_temperaturas
```

```python
df_cidades = pd.DataFrame({
    "nome": ["Ana", "Bruno", "Carlos", "Daniele", "Val"],
    "cidade": ["São Paulo", np.nan, "Curitiba", np.nan, "Belém"]
})

df_cidades["cidade_preenchida"] = df_cidades["cidade"].fillna("não preenchido")

df_cidades
```
